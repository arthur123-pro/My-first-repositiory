```python
regional_spread= population_data.groupby("continent")["2023 population"].sum().reset_index()
plt.figure(figsize=(8,8))
plt.pie(regional_spread["2023 population"],labels=regional_spread['continent'],autopct='%1.1f%%',startangle=140,colors=sns.color_palette('Set2'))
plt.title("How the Global Population is Shared across the Regions in 2023",fontsize=15,pad=22)
plt.show()
```

This pie chart helps us break down the world population across distinct Regions.
Here each slice represents a continent's percentage share of the total world population in 2023. And if you look kinly, it clearly answers our question 2 by showing that the world's population is notshared evenly across the globe.  Showing that Asia holds then highest percentage of the world's population at 59.1% almost 60%, then Africa. ultimately, continents like Oceania and America hold the smallest percentage of the world's population and this visually proves that human settlement is heavly concentrated in specific regions rather than just being evenly spreadout.

```python
population_data = pd.read_csv("world_population_data.csv")
population_data.head()
```

```python

sns.histplot(data=population_data,x="2023 population",kde=True,color="red",edgecolor="yellow",bins=20,log_scale=True)

plt.title("The Global Distribution of Country Populations in 2023 ")
plt.xlabel("2023 Population Size (Log Scale)")
plt.ylabel("Number of Countries")
plt.show()
```

This Histogram answers our first question simply because it shows the distribution of national populations across logarithimic scale.

If we look at the x-axis, it breaks down the country populations into groups that are scaled by the powers of 10. And when we look at the y-axis,it displays to us the number of countries that fall into different groups. So we found out that by using the log scale, we can clearly visualize the massive gap in a global wide-spread populations.

Our chart displays a bell-shaped  spreed, showing that the majority of the countries are comfortably between one million and a hundred million people. The other part pointing the 10e9 is where the countries with more than 100 million people lie like indian among others which are the outliers for this case.



This is a historical line chart that tracks the growth speeds of total combined populations across different global regions from 1970 to 2023.
If we look right at the x-axis,it shows the timeline progression over more than 30 years,showing specific milestones between 1970 and 2023.  And then the y-axis,  tells us of the total population counts scaled in billions. Now this chart clearly answers our forth question by showing how widely different the population rate increases across these Regions. As shown on the line chart, Asia has been increasing aggressively over the years ,though its shape increase started lowering as it approached 2023 not as sharp as it was in the previous years.

```python
density = [column for column in population_data.columns if 'dens' in column.lower()][0]
avg_density = population_data.groupby('continent')[density].mean().sort_values(ascending=False).reset_index()
plt.figure(figsize=(10, 5))
sns.barplot(data=avg_density, x='continent',y=density,edgecolor='black')
plt.title("Average Crowding Levels (Densities) Across Continents",fontsize=10,pad=13)
plt.xlabel("Continent", fontsize=10)
plt.ylabel("Average People per km²",fontsize=10)
plt.grid(axis='y',linestyle="--",alpha=0.5)

plt.tight_layout()
plt.show()
```

This bar graph displays the crowding population density globally. 
The x-axis, lists the different continents sorted from that most crowded to the least crowded ones.
The y-axis then shows us the average number of people living within one square kilometer.showing as that Asia is the most crowded region and the Oceania is the least crowded region which answers our third question.
